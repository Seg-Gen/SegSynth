**Block fold 1. One arm per session. Change `ARM` in section 2 and run top to bottom.**

| arm | train pairs | composition |
|---|---|---|
| `R0` | 804 | 804 real fold-1 train pairs |
| `R1` | 1608 | the same 804 real + 804 synthetic from `fake_pairs_R1/` |

Validation is the **136 real fold-1 tiles (408 images) in both arms**, never a synthetic pixel.

This is the second architecture through the Stage-4 harness. Everything outside the model is
identical to `geonusaf-stage4-segformer-b0-R0R1.ipynb` — same sections, same split assert, same
synthetic loader, same step budget, same paired bootstrap — so the two models' Stage-4 tables are
directly comparable and the result JSONs concatenate without reconciliation.

## Three deliberate departures from `unetformer_final_v2.ipynb`

**1. Step-based, not epoch-based.** R1's training set is twice R0's, so a pass is 100 steps
against R0's 50. Holding an epoch count fixed would hand R1 twice the gradient updates and the
result would measure compute, not data. Both arms get `MAX_STEPS = 6000` with a 500-step warmup
and a cosine horizon of 6000 — which at R0 is *exactly* the old 120-epoch schedule
(120 x 50 = 6000). Validation every 200 steps, patience 8 validations. The old `MAX_EPOCHS = 200`
ceiling is gone: the cosine is already flat at zero past step 6000, so the extra epochs it allowed
never changed anything except wall clock.

**2. Augmentation moves to the worker-local RNG.** The part-1 UNetFormer notebook already avoided
the frozen-augmentation bug, but it did so by seeding a fresh `RandomState` from `self.epoch` and
forcing `persistent_workers=False` so the workers would see each new epoch. A step-based loop has
no epoch boundary to bump, so that mechanism cannot carry over. Draws here come from the
worker-local `np.random` stream, which advances on every call regardless of worker lifetime —
which also lets `persistent_workers=True` come back and removes a worker respawn per pass.
Section 11 asserts liveness.

**3. Deterministic algorithms stay off** — as in part 1, where `DETERMINISTIC = False` was already
the default. Recorded here only because the SegFormer-B0 Stage-4 notebook had to change it, and
the two harnesses must be stated as matched.

What is **not** changed: the architecture, `DECODE_CHANNELS = 64`, window 8, 8 heads, aux weight
0.4, drop path 0.1, LR 3e-4 / 3e-5, AdamW wd 0.01, EMA 0.999, label smoothing 0.05, and the
weighted-CE + Lovasz loss. The model is the part-1 model, byte for byte, and `arch_sig` asserts it.

Consequence: **the R0 number this notebook produces will not equal the published part-1 fold-1
UNetFormer result**, because that run used an epoch ceiling, epoch-granularity early stopping and
epoch-seeded augmentation. That is the point. R0 and R1 must come out of the same harness or the
delta means nothing. Do not mix the old fold-1 number into the Stage-4 table.

## 2. Configuration

`ARM` is the only field you change between sessions. Everything else is identical across the two

The model block is copied unchanged from `unetformer_final_v2.ipynb`, including `LR_PAPER`, which

## 3. Determinism — deliberately OFF

`torch.use_deterministic_algorithms(True)` forces a sort-based scatter backward that costs roughly
6x on this workload, and it buys nothing here: both arms share the seed, the split and the
schedule, so run-to-run kernel nondeterminism is common-mode noise on a comparison, not a
confound. `cudnn.benchmark` is on instead, since the input shape never varies.

This matches the part-1 UNetFormer notebook, where `DETERMINISTIC` was already `False` — so unlike
the SegFormer-B0 arm pair, nothing about the reported efficiency numbers changes here.

```python
def set_seed(seed: int):
    random.seed(seed); np.random.seed(seed)
    torch.manual_seed(seed); torch.cuda.manual_seed_all(seed)
    os.environ["PYTHONHASHSEED"] = str(seed)
    torch.backends.cudnn.deterministic = False
    torch.backends.cudnn.benchmark = True          # fixed 512x512 shapes -> autotune pays off
    torch.use_deterministic_algorithms(False)

def worker_init_fn(worker_id):
    # each worker gets its OWN np.random stream, and that stream advances on every augmentation
    # draw. This is what makes augmentation live across passes -- see section 11.
    s = (cfg.SEED * 7919 + worker_id) % (2**31)
    np.random.seed(s); random.seed(s)

set_seed(cfg.SEED)
g = torch.Generator(); g.manual_seed(cfg.SEED)
print("seeded:", cfg.SEED, "| deterministic algorithms: OFF | cudnn.benchmark: ON")
```

Verbatim from the part-1 baseline, including the 1224 / 408 asserts. `DATA_ROOT` is auto-detected

```python
def find_data_root(explicit):
    if explicit and Path(explicit, "images").is_dir(): return explicit
    for p in Path("/kaggle/input").rglob("satellite_dataset"):
        if (p / "images").is_dir() and (p / "masks").is_dir(): return str(p)
    raise FileNotFoundError("satellite_dataset not found -- attach sugam24/satellite-datasets")

DATA_ROOT = find_data_root(cfg.DATA_ROOT)
IMG_DIR  = Path(DATA_ROOT) / "images"
MASK_DIR = Path(DATA_ROOT) / "masks"
print("DATA_ROOT:", DATA_ROOT)

PAT = re.compile(r"^image(\d+)(hf|vf)?\.png$")
records = []          # (base_id:int, variant:str, image_name:str, mask_name:str)
for f in sorted(os.listdir(IMG_DIR)):
    m = PAT.match(f)
    assert m, f"unparsed image filename: {f}"
    base_id, variant = int(m.group(1)), (m.group(2) or "orig")
    mask_name = "mask" + f[len("image"):]
    assert (MASK_DIR / mask_name).exists(), f"missing mask for {f}"
    records.append((base_id, variant, f, mask_name))

groups: Dict[int, List[tuple]] = {}
for r in records:
    groups.setdefault(r[0], []).append(r)

BASE_IDS = sorted(groups)
assert len(records) == 1224, len(records)
assert len(BASE_IDS) == 408, len(BASE_IDS)
assert all(len(v) == 3 for v in groups.values())
print(f"{len(records)} files | {len(BASE_IDS)} base tiles | id range {BASE_IDS[0]}..{BASE_IDS[-1]}")
```

## 5. Mask colour diagnostic

The channel-order guard. If BGR/RGB ever inverted, Residential and River would swap and every
number below would be wrong in a way that looks plausible.

```python
BGR_TO_CLASS = {
    (128, 0, 0): 0, (0, 128, 0): 1, (0, 0, 128): 2,
    (0, 128, 128): 3, (128, 128, 0): 4, (128, 0, 128): 5,
}

def pack_bgr(a: np.ndarray) -> np.ndarray:
    a = a.astype(np.uint32)
    return (a[..., 0] << 16) | (a[..., 1] << 8) | a[..., 2]

LUT = np.full(1 << 24, 255, dtype=np.uint8)
for bgr, idx in BGR_TO_CLASS.items():
    LUT[pack_bgr(np.array(bgr, dtype=np.uint8).reshape(1, 1, 3))[0, 0]] = idx

rng = np.random.RandomState(cfg.SEED)
sample = [records[i] for i in rng.choice(len(records), 40, replace=False)]
class_px = np.zeros(7, dtype=np.int64)
for _, _, _, mname in sample:
    m = cv2.imread(str(MASK_DIR / mname), cv2.IMREAD_COLOR)
    idx = LUT[pack_bgr(m)]
    class_px += np.bincount(np.where(idx == 255, 6, idx).ravel(), minlength=7)

tot = class_px.sum()
for i, nm in enumerate(list(cfg.CLASS_NAMES) + ["<ignore>"]):
    print(f"  {i if i < 6 else 255:>3} {nm:<14} {class_px[i]/tot*100:6.2f}%")
assert class_px[0] > class_px[2] * 3, "Residential <= River -- channel order inverted. STOP."
print("\nchannel-order check PASSED")
```

## 6. One-time repack to a 512 cache

Unchanged. If you already saved this cache as a Kaggle dataset from a part-1 run, point
`PREBUILT_CACHE` at it and this becomes a no-op — it saves ~15 min per session and there are two
sessions here.

```python
from concurrent.futures import ProcessPoolExecutor

CACHE = Path(cfg.PREBUILT_CACHE) if cfg.PREBUILT_CACHE else Path(cfg.CACHE_DIR)
MARKER = CACHE / "_complete.json"

def _repack_one(rec):
    _, _, iname, mname = rec
    img = cv2.imread(str(IMG_DIR / iname), cv2.IMREAD_COLOR)
    img = cv2.resize(img, (cfg.IMG_SIZE, cfg.IMG_SIZE), interpolation=cv2.INTER_AREA)
    msk = cv2.imread(str(MASK_DIR / mname), cv2.IMREAD_COLOR)
    lab = LUT[pack_bgr(msk)]
    lab = cv2.resize(lab, (cfg.IMG_SIZE, cfg.IMG_SIZE), interpolation=cv2.INTER_NEAREST)
    cv2.imwrite(str(CACHE / "images" / iname), img, [cv2.IMWRITE_PNG_COMPRESSION, 3])
    cv2.imwrite(str(CACHE / "labels" / mname), lab, [cv2.IMWRITE_PNG_COMPRESSION, 3])
    return np.bincount(np.where(lab == 255, 6, lab).ravel(), minlength=7)

if MARKER.exists():
    print(f"cache present at {CACHE} ({json.load(open(MARKER))['n']} pairs) -- skipping repack")
else:
    assert not cfg.PREBUILT_CACHE, f"PREBUILT_CACHE set but {MARKER} missing"
    (CACHE / "images").mkdir(parents=True, exist_ok=True)
    (CACHE / "labels").mkdir(parents=True, exist_ok=True)
    t0 = time.time(); counts = {}
    with ProcessPoolExecutor(max_workers=4) as ex:
        for rec, b in zip(records, ex.map(_repack_one, records, chunksize=8)):
            counts[rec[2]] = b.tolist()
            if len(counts) % 200 == 0:
                el = time.time() - t0
                print(f"  {len(counts)}/{len(records)}  {el:.0f}s  "
                      f"eta {el/len(counts)*(len(records)-len(counts)):.0f}s")
    json.dump({"n": len(counts), "counts": counts}, open(MARKER, "w"))
    print(f"repacked {len(counts)} pairs in {time.time()-t0:.0f}s")

PIX_COUNTS = {k: np.array(v, dtype=np.int64) for k, v in json.load(open(MARKER))["counts"].items()}
```

## 8. Split, and the four-way agreement with the generators

`block_split` is verbatim from part 1 and from Stages 1-3. The sha1 over the sorted train ids is
**asserted**, not printed for eyeballing: Stage 1, Stage 2 and Stage 3 all trained or sampled on
this exact list, and if Stage 4's real half were a different split the synthetic pairs would have
been generated from tiles that are now in validation. That is the leakage failure this experiment
cannot survive, and it would be invisible in every metric.

The same constant is asserted in the SegFormer-B0 Stage-4 notebook, so all five arms across the
two models train on one identical real split.

```python
def block_split(base_ids: List[int], fold: int, n_folds: int, buffer: int):
    n = len(base_ids)
    edges = [round(i * n / n_folds) for i in range(n_folds + 1)]
    lo, hi = edges[fold], edges[fold + 1]
    val = base_ids[lo:hi]
    train = base_ids[:max(0, lo - buffer)] + base_ids[min(n, hi + buffer):]
    return train, val

tr_ids, va_ids = block_split(BASE_IDS, cfg.FOLD, cfg.N_FOLDS, cfg.BLOCK_BUFFER)

def expand(ids): return [r for i in ids for r in groups[i]]
real_tr_recs, va_recs = expand(tr_ids), expand(va_ids)

assert not (set(tr_ids) & set(va_ids)), "base-tile leakage between train and val"
split_sha1 = hashlib.sha1(",".join(map(str, sorted(tr_ids))).encode()).hexdigest()
assert split_sha1 == cfg.UPSTREAM_TRAIN_SHA1, (
    "SPLIT DISAGREEMENT with the generators -- the synthetic pairs were made from a different "
    f"train set.\n  upstream: {cfg.UPSTREAM_TRAIN_SHA1}\n  here    : {split_sha1}")

print(f"{cfg.SPLIT_MODE} fold {cfg.FOLD}: {len(tr_ids)} train tiles / {len(va_ids)} val tiles")
print(f"  -> {len(real_tr_recs)} real train pairs / {len(va_recs)} val images")
print(f"  train_sha1 {split_sha1[:12]} -- four-way agreement with S1/S2/S3 ASSERTED")
assert len(real_tr_recs) == 804, len(real_tr_recs)
```

## 9. The synthetic half

Loaded only when `ARM == "R1"`. The Stage-3 contract is re-asserted here rather than trusted:
804 pairs, indices 0..803 complete, images 512x512x3 `uint8` BGR, masks 512x512 `uint8` with
values in {0..5, 255}. Same dtypes and channel order as the real cache, so they go into the same
`RAM` dict and the Dataset cannot tell them apart.

Synthetic records carry `base_id = -(i+1)`. Negative ids can never collide with a real tile, which
means the assert at the end of this section — that no synthetic record reached validation — is
checking something real rather than restating an assumption.

The printed class-share table is worth reading before you train. The synthetic half carries River
in roughly twice as many masks as the real half does; that is the intended effect and it is also
the thing most likely to shift the confusion matrix.

```python
def find_fake_root(explicit, name="fake_pairs_R1"):
    if explicit and Path(explicit).is_dir(): return explicit
    for p in Path("/kaggle/input").rglob(name):
        if p.is_dir() and any(p.glob("genimage*.png")): return str(p)
    raise FileNotFoundError(f"{name} not found -- attach sugam24/geonusaf-fake-pairs-block-fold1")

fake_tr_recs = []
if USE_FAKE:
    FAKE_ROOT = Path(find_fake_root(cfg.FAKE_ROOT))
    print("FAKE_ROOT:", FAKE_ROOT)

    idxs = sorted(int(p.stem[len("genimage"):]) for p in FAKE_ROOT.glob("genimage*.png"))
    assert len(idxs) == cfg.N_FAKE, f"{len(idxs)} synthetic images, expected {cfg.N_FAKE}"
    assert idxs == list(range(cfg.N_FAKE)), "synthetic index range is not 0..n-1"

    t0 = time.time(); fake_px = np.zeros(7, dtype=np.int64)
    for i in idxs:
        iname, mname = f"genimage{i}.png", f"genmask{i}.png"
        img = cv2.imread(str(FAKE_ROOT / iname), cv2.IMREAD_COLOR)
        lab = cv2.imread(str(FAKE_ROOT / mname), cv2.IMREAD_GRAYSCALE)
        assert img is not None and lab is not None, f"unreadable synthetic pair {i}"
        assert img.shape == (cfg.IMG_SIZE, cfg.IMG_SIZE, 3) and img.dtype == np.uint8, \
            f"synthetic image {i} is {img.shape} {img.dtype}"
        assert lab.shape == (cfg.IMG_SIZE, cfg.IMG_SIZE) and lab.dtype == np.uint8, \
            f"synthetic mask {i} is {lab.shape} {lab.dtype}"
        b = np.bincount(np.where(lab == 255, 6, lab).ravel(), minlength=7)
        assert b.shape[0] == 7, f"synthetic mask {i} has values outside the contract"
        RAM[iname] = (img, lab)
        PIX_COUNTS[iname] = b
        fake_px += b
        fake_tr_recs.append((-(i + 1), "fake", iname, mname))
        if (i + 1) % 200 == 0: print(f"  loaded {i+1}/{cfg.N_FAKE}")
    print(f"loaded {len(fake_tr_recs)} synthetic pairs in {time.time()-t0:.0f}s")

    real_px = np.zeros(7, dtype=np.int64)
    for _, _, iname, _ in real_tr_recs: real_px += PIX_COUNTS[iname]
    print(f"\n{'class':<14}{'real train %':>14}{'synthetic %':>13}{'ratio':>8}")
    for i, nm in enumerate(list(cfg.CLASS_NAMES) + ["<ignore>"]):
        r_, f_ = real_px[i]/real_px.sum()*100, fake_px[i]/fake_px.sum()*100
        print(f"{nm:<14}{r_:13.3f}%{f_:12.3f}%{(f_/max(r_,1e-9)):8.2f}")

tr_recs = real_tr_recs + fake_tr_recs
FAKE_NAMES = {r[2] for r in fake_tr_recs}
assert not (FAKE_NAMES & {r[2] for r in va_recs}), "SYNTHETIC PAIR IN VALIDATION -- stop"
assert all(r[0] > 0 for r in va_recs), "negative base id in validation"
print(f"\narm {cfg.ARM}: {len(tr_recs)} train pairs "
      f"({len(real_tr_recs)} real + {len(fake_tr_recs)} synthetic) | {len(va_recs)} val images")
```

## 10. Effective-number class weights

Computed from the **real** train pairs only, so `CLASS_W` is bit-identical in both arms and the
loss is not a second moving variable. `WEIGHTS_FROM="mixed"` recomputes over real+synthetic
instead; both are printed either way so the paper can state what the choice cost.

```python
EN_BETA_AUTO = True

def effective_number_weights(recs, cap, n_classes, beta_fixed, auto):
    n = np.zeros(n_classes, dtype=np.float64)
    for _, _, iname, _ in recs:
        n += PIX_COUNTS[iname][:n_classes]
    beta = (1.0 - 1.0 / n.sum()) if auto else beta_fixed
    eff = (1.0 - np.power(beta, n)) / (1.0 - beta)
    w = 1.0 / np.maximum(eff, 1e-12)
    w = w / w.mean()
    w = np.minimum(w, cap)
    return w / w.mean(), n, beta

W_REAL, PX_REAL, BETA_R = effective_number_weights(
    real_tr_recs, cfg.WEIGHT_CAP, cfg.N_CLASSES, cfg.EN_BETA, EN_BETA_AUTO)
W_MIXED, PX_MIXED, BETA_M = effective_number_weights(
    tr_recs, cfg.WEIGHT_CAP, cfg.N_CLASSES, cfg.EN_BETA, EN_BETA_AUTO)

CLASS_W = W_REAL if cfg.WEIGHTS_FROM == "real" else W_MIXED
print(f"WEIGHTS_FROM = {cfg.WEIGHTS_FROM}\n")
print(f"{'class':<14}{'train px %':>12}{'w(real)':>10}{'w(mixed)':>11}")
for i, nm in enumerate(cfg.CLASS_NAMES):
    print(f"{nm:<14}{PX_REAL[i]/PX_REAL.sum()*100:11.2f}%{W_REAL[i]:10.3f}{W_MIXED[i]:11.3f}")
print(f"weight ratio max/min: real {W_REAL.max()/W_REAL.min():.1f} | "
      f"mixed {W_MIXED.max()/W_MIXED.min():.1f}")

CLASS_W_T = torch.tensor(CLASS_W, dtype=torch.float32).cuda()
```

## 11. Dataset and DataLoaders — augmentation on the worker-local RNG

The part-1 UNetFormer notebook seeded a fresh `RandomState` from `self.epoch` and kept
`persistent_workers=False` so each new epoch would reach the workers. There are no epochs here, so
that mechanism has nothing to key on. Draws now come from the worker-local `np.random` stream,
which `worker_init_fn` seeds once per worker and which then advances on every single call — no
bookkeeping to propagate, nothing to go stale, and `persistent_workers=True` is safe again.

The assert below draws the same record twice and requires the results to differ. It runs in the
main process, so it tests the mechanism rather than the multiprocessing path, but the mechanism is
exactly what has to be live.

No hflip/vflip: those are already materialised as separate rows in the 1224 real files. Synthetic
and real records are read through the same path; `AUG_FLIP_FAKE` is the one place they can differ,
and it is off by default — see section 2.

```python
IMNET_MEAN = np.array([0.485, 0.456, 0.406], np.float32)
IMNET_STD  = np.array([0.229, 0.224, 0.225], np.float32)

class SatSeg(Dataset):
    def __init__(self, recs, train: bool):
        self.recs, self.train = recs, train

    def __len__(self): return len(self.recs)

    def _read(self, iname, mname):
        if cfg.PRELOAD_RAM:
            img, lab = RAM[iname]
            return img.copy(), lab.copy()
        if iname in FAKE_NAMES:
            img = cv2.imread(str(FAKE_ROOT / iname), cv2.IMREAD_COLOR)
            lab = cv2.imread(str(FAKE_ROOT / mname), cv2.IMREAD_GRAYSCALE)
        else:
            img = cv2.imread(str(CACHE / "images" / iname), cv2.IMREAD_COLOR)
            lab = cv2.imread(str(CACHE / "labels" / mname), cv2.IMREAD_GRAYSCALE)
        return img, lab

    def __getitem__(self, i):
        _, variant, iname, mname = self.recs[i]
        img, lab = self._read(iname, mname)

        if self.train:
            if cfg.AUG_FLIP_FAKE and variant == "fake":
                # real records carry hf/vf as separate rows; this gives synthetic ones the same
                # flip coverage instead of leaving them with only rot90
                f = np.random.randint(3)
                if f == 1: img, lab = img[:, ::-1], lab[:, ::-1]
                elif f == 2: img, lab = img[::-1], lab[::-1]
            if cfg.AUG_ROT90:
                k = np.random.randint(4)
                if k: img, lab = np.rot90(img, k), np.rot90(lab, k)
            img = np.ascontiguousarray(img); lab = np.ascontiguousarray(lab)
            if cfg.AUG_PHOTOMETRIC:
                a = 1.0 + np.random.uniform(-cfg.AUG_CONTRAST, cfg.AUG_CONTRAST)
                b = 255.0 * np.random.uniform(-cfg.AUG_BRIGHTNESS, cfg.AUG_BRIGHTNESS)
                img = np.clip(img.astype(np.float32) * a + b, 0, 255).astype(np.uint8)

        x = cv2.cvtColor(img, cv2.COLOR_BGR2RGB).astype(np.float32) / 255.0
        x = (x - IMNET_MEAN) / IMNET_STD
        x = torch.from_numpy(x.transpose(2, 0, 1))
        y = torch.from_numpy(np.ascontiguousarray(lab).astype(np.int64))
        return x, y

train_ds, val_ds = SatSeg(tr_recs, True), SatSeg(va_recs, False)

# the liveness check: two draws of the same record must differ
np.random.seed(0)
_a, _ = train_ds[0]
_b, _ = train_ds[0]
assert not torch.allclose(_a, _b), (
    "augmentation is STATIC -- two draws of record 0 are identical. A step-based loop has no "
    "epoch counter to fall back on, so this must come from the worker-local stream.")
print("augmentation liveness check PASSED (two draws of record 0 differ)")

train_ld = DataLoader(train_ds, batch_size=cfg.BATCH_SIZE, shuffle=True, drop_last=True,
                      num_workers=cfg.NUM_WORKERS, pin_memory=True, persistent_workers=True,
                      worker_init_fn=worker_init_fn, generator=g)
val_ld   = DataLoader(val_ds, batch_size=cfg.BATCH_SIZE, shuffle=False,
                      num_workers=2, pin_memory=True, persistent_workers=True,
                      worker_init_fn=worker_init_fn)

steps_per_pass = len(train_ld)
xb, yb = next(iter(train_ld))
print(f"{xb.shape} {xb.dtype} | {yb.shape} {yb.dtype}")
print(f"steps per pass: {steps_per_pass} | {cfg.MAX_STEPS} steps = "
      f"{cfg.MAX_STEPS/steps_per_pass:.1f} passes over the training set")
```

## 12. Loss — weighted CE + Lovasz-Softmax, plus deep supervision

Unchanged from part 1, including the fp32 cast outside autocast and the present-classes-only
Lovasz. `CLASS_W_T` comes from section 10 and is identical in both arms.

UNetFormer's auxiliary head is supervised with the **same** combined loss at weight
`AUX_WEIGHT = 0.4`, applied in section 18 rather than inside `ComboLoss`, so that the logged
`train_ce` and `train_lov` stay comparable with the other four models' notebooks. The aux term is
logged separately for the same reason.

```python
def lovasz_grad(gt_sorted):
    p = len(gt_sorted); gts = gt_sorted.sum()
    inter = gts - gt_sorted.float().cumsum(0)
    union = gts + (1 - gt_sorted).float().cumsum(0)
    jac = 1.0 - inter / union
    if p > 1: jac[1:p] = jac[1:p] - jac[0:-1]
    return jac

def lovasz_softmax_flat(probas, labels, n_classes):
    if probas.numel() == 0: return probas.sum() * 0.0
    losses = []
    for c in range(n_classes):
        fg = (labels == c).float()
        if fg.sum() == 0: continue                     # "present" classes only
        err = (fg - probas[:, c]).abs()
        err_sorted, perm = torch.sort(err, 0, descending=True)
        losses.append(torch.dot(err_sorted, lovasz_grad(fg[perm])))
    return torch.stack(losses).mean() if losses else probas.sum() * 0.0

class ComboLoss(nn.Module):
    def __init__(self, weights, lam, ignore, n_classes, smooth=0.0):
        super().__init__()
        self.ce = nn.CrossEntropyLoss(weight=weights, ignore_index=ignore, label_smoothing=smooth)
        self.lam, self.ignore, self.nc = lam, ignore, n_classes

    def forward(self, logits, target):
        logits = logits.float()                        # fp32 regardless of autocast
        ce = self.ce(logits, target)
        B, C, H, W = logits.shape
        p = logits.permute(0, 2, 3, 1).reshape(-1, C).softmax(1)
        t = target.reshape(-1)
        keep = t != self.ignore
        lv = lovasz_softmax_flat(p[keep], t[keep], self.nc)
        return ce + self.lam * lv, ce.detach(), lv.detach()

criterion = ComboLoss(CLASS_W_T, cfg.LAMBDA_LOVASZ, cfg.IGNORE_INDEX,
                      cfg.N_CLASSES, cfg.LABEL_SMOOTHING).cuda()
print(f"loss ready | lambda {cfg.LAMBDA_LOVASZ} | label smoothing {cfg.LABEL_SMOOTHING} "
      f"| aux weight {cfg.AUX_WEIGHT}")
```

## 13. Model, optimizer, step-based scheduler

The architecture is carried verbatim from `unetformer_final_v2.ipynb` — an independent
implementation of Wang et al. (2022), because the reference repo (`WangLibo1995/GeoSeg`) is
GPL-3.0 and copying it would put the paper's code release under GPL-3.0 too. `arch_signature`
hashes the state-dict keys and shapes; it is stamped into every checkpoint and verified before any
load, so a checkpoint from a different architecture can never be half-loaded into this run.

The only change from part 1 is the scheduler horizon: `total_steps` is now `COS_HORIZON_STEPS`
straight from the config rather than `steps_per_epoch * COS_HORIZON_EPOCHS`. That single line is
what makes the two arms compute-matched — under the old form, R1's doubled `steps_per_epoch` would
have stretched the cosine to 12000 steps and given it twice the schedule.

`LR_PAPER = 6e-4` is recorded and unused; the run is at 3e-4 / 3e-5, as in part 1.

```python
from einops import rearrange

# ---------------------------------------------------------------- building blocks
class Conv(nn.Sequential):
    def __init__(self, i, o, k=3, stride=1, dilation=1, bias=False):
        super().__init__(nn.Conv2d(i, o, k, stride=stride, dilation=dilation,
                                   padding=((stride - 1) + dilation * (k - 1)) // 2, bias=bias))

class ConvBN(nn.Sequential):
    def __init__(self, i, o, k=3, stride=1, dilation=1):
        super().__init__(nn.Conv2d(i, o, k, stride=stride, dilation=dilation,
                                   padding=((stride - 1) + dilation * (k - 1)) // 2, bias=False),
                         nn.BatchNorm2d(o))

class ConvBNReLU(nn.Sequential):
    def __init__(self, i, o, k=3, stride=1, dilation=1):
        super().__init__(nn.Conv2d(i, o, k, stride=stride, dilation=dilation,
                                   padding=((stride - 1) + dilation * (k - 1)) // 2, bias=False),
                         nn.BatchNorm2d(o), nn.ReLU6(inplace=True))

class SeparableConvBN(nn.Sequential):
    def __init__(self, i, o, k=3, stride=1, dilation=1):
        super().__init__(nn.Conv2d(i, i, k, stride=stride, dilation=dilation,
                                   padding=((stride - 1) + dilation * (k - 1)) // 2, groups=i, bias=False),
                         nn.Conv2d(i, o, 1, bias=False), nn.BatchNorm2d(o))

class DropPath(nn.Module):
    # Stochastic depth on a residual branch. No parameters, no state_dict keys.
    def __init__(self, p=0.0):
        super().__init__(); self.p = float(p)
    def forward(self, x):
        if self.p == 0.0 or not self.training: return x
        keep = 1.0 - self.p
        mask = x.new_empty((x.shape[0],) + (1,) * (x.ndim - 1)).bernoulli_(keep)
        return x * mask / keep

class Mlp(nn.Module):
    def __init__(self, dim, hidden, drop=0.0):
        super().__init__()
        self.fc1, self.act = nn.Conv2d(dim, hidden, 1), nn.ReLU6(inplace=True)
        self.fc2, self.drop = nn.Conv2d(hidden, dim, 1), nn.Dropout(drop)
    def forward(self, x):
        return self.drop(self.fc2(self.drop(self.act(self.fc1(x)))))

# ---------------------------------------------------------------- global-local attention
class GlobalLocalAttention(nn.Module):
    # Windowed MHSA (global branch) in parallel with a conv branch (local), plus pooled cross context.
    def __init__(self, dim=64, num_heads=8, window_size=8, qkv_bias=False):
        super().__init__()
        assert dim % num_heads == 0
        self.nh, self.ws = num_heads, window_size
        self.scale = (dim // num_heads) ** -0.5
        self.qkv = Conv(dim, 3 * dim, k=1, bias=qkv_bias)
        self.local1, self.local2 = ConvBN(dim, dim, k=3), ConvBN(dim, dim, k=1)
        self.proj = SeparableConvBN(dim, dim, k=window_size)
        # kernel (ws,1) with padding ws//2-1 shrinks that axis by exactly 1; the pad before proj restores it
        self.attn_x = nn.AvgPool2d((window_size, 1), stride=1, padding=(window_size // 2 - 1, 0))
        self.attn_y = nn.AvgPool2d((1, window_size), stride=1, padding=(0, window_size // 2 - 1))
        ws = window_size
        self.rpb_table = nn.Parameter(torch.zeros((2 * ws - 1) ** 2, num_heads))
        nn.init.trunc_normal_(self.rpb_table, std=0.02)
        c = torch.stack(torch.meshgrid(torch.arange(ws), torch.arange(ws), indexing="ij")).flatten(1)
        rel = (c[:, :, None] - c[:, None, :]).permute(1, 2, 0).contiguous()
        rel[:, :, 0] += ws - 1; rel[:, :, 1] += ws - 1; rel[:, :, 0] *= 2 * ws - 1
        self.register_buffer("rpi", rel.sum(-1), persistent=False)

    def forward(self, x):
        B, C, H, W = x.shape
        local = self.local2(x) + self.local1(x)
        ws = self.ws
        ph, pw = (ws - H % ws) % ws, (ws - W % ws) % ws
        xp = F.pad(x, (0, pw, 0, ph), mode="reflect") if (ph or pw) else x
        Hp, Wp = xp.shape[-2:]
        q, k, v = rearrange(self.qkv(xp),
                            "b (qkv h d) (hh w1) (ww w2) -> qkv (b hh ww) h (w1 w2) d",
                            qkv=3, h=self.nh, w1=ws, w2=ws)
        bias = self.rpb_table[self.rpi.view(-1)].view(ws * ws, ws * ws, self.nh).permute(2, 0, 1)
        attn = ((q @ k.transpose(-2, -1)) * self.scale + bias.unsqueeze(0)).softmax(-1) @ v
        attn = rearrange(attn, "(b hh ww) h (w1 w2) d -> b (h d) (hh w1) (ww w2)",
                         b=B, hh=Hp // ws, ww=Wp // ws, w1=ws, w2=ws)[:, :, :H, :W]
        fused = attn + local
        cross = (self.attn_x(F.pad(fused, (0, 0, 0, 1), mode="reflect")) +
                 self.attn_y(F.pad(fused, (0, 1, 0, 0), mode="reflect")))
        return self.proj(F.pad(fused + cross, (0, 1, 0, 1), mode="reflect"))[:, :, :H, :W]

class Block(nn.Module):
    # Global-Local Transformer Block. BatchNorm rather than LayerNorm, as in the paper.
    def __init__(self, dim=64, num_heads=8, window_size=8, mlp_ratio=4.0, drop=0.0, drop_path=0.0):
        super().__init__()
        self.norm1, self.norm2 = nn.BatchNorm2d(dim), nn.BatchNorm2d(dim)
        self.attn = GlobalLocalAttention(dim, num_heads, window_size)
        self.mlp = Mlp(dim, int(dim * mlp_ratio), drop)
        self.dp1, self.dp2 = DropPath(drop_path), DropPath(drop_path)
    def forward(self, x):
        x = x + self.dp1(self.attn(self.norm1(x)))
        return x + self.dp2(self.mlp(self.norm2(x)))

# ---------------------------------------------------------------- fusion / heads
class WF(nn.Module):
    # Weighted fusion of an upsampled decoder feature with an encoder skip; the weights are learned.
    def __init__(self, in_ch, dc, eps=1e-8):
        super().__init__()
        self.pre_conv, self.post_conv = Conv(in_ch, dc, k=1), ConvBNReLU(dc, dc, k=3)
        self.weights, self.eps = nn.Parameter(torch.ones(2, dtype=torch.float32)), eps
    def forward(self, x, res):
        x = F.interpolate(x, size=res.shape[-2:], mode="bilinear", align_corners=False)
        w = torch.relu(self.weights); w = w / (w.sum() + self.eps)
        return self.post_conv(w[0] * self.pre_conv(res) + w[1] * x)

class FeatureRefinementHead(nn.Module):
    def __init__(self, in_ch, dc, eps=1e-8):
        super().__init__()
        self.pre_conv, self.post_conv = Conv(in_ch, dc, k=1), ConvBNReLU(dc, dc, k=3)
        self.weights, self.eps = nn.Parameter(torch.ones(2, dtype=torch.float32)), eps
        self.pa = nn.Sequential(nn.Conv2d(dc, dc, 3, padding=1, groups=dc), nn.Sigmoid())
        self.ca = nn.Sequential(nn.AdaptiveAvgPool2d(1), Conv(dc, max(dc // 16, 4), k=1), nn.ReLU6(True),
                                Conv(max(dc // 16, 4), dc, k=1), nn.Sigmoid())
        self.shortcut, self.proj = ConvBN(dc, dc, k=1), SeparableConvBN(dc, dc, k=3)
        self.act = nn.ReLU6(True)
    def forward(self, x, res):
        x = F.interpolate(x, size=res.shape[-2:], mode="bilinear", align_corners=False)
        w = torch.relu(self.weights); w = w / (w.sum() + self.eps)
        x = self.post_conv(w[0] * self.pre_conv(res) + w[1] * x)
        sc = self.shortcut(x)
        x = self.pa(x) * x + self.ca(x) * x
        return self.act(self.proj(x) + sc)

class AuxHead(nn.Module):
    def __init__(self, dc, n_cls):
        super().__init__()
        self.body = nn.Sequential(ConvBNReLU(dc, dc), nn.Dropout2d(0.1), Conv(dc, n_cls, k=1))
    def forward(self, x): return self.body(x)

class UNetFormer(nn.Module):
    def __init__(self, encoder="resnet18", n_classes=6, dc=64, window_size=8,
                 num_heads=8, dropout=0.1, drop_path=0.1, pretrained=True):
        super().__init__()
        self.backbone = timm.create_model(encoder, features_only=True,
                                          out_indices=(1, 2, 3, 4), pretrained=pretrained)
        ec = self.backbone.feature_info.channels()          # (64, 128, 256, 512) at /4 /8 /16 /32
        self.pre_conv = ConvBN(ec[3], dc, k=1)
        self.b4 = Block(dc, num_heads, window_size, drop_path=drop_path)
        self.p3, self.b3 = WF(ec[2], dc), Block(dc, num_heads, window_size, drop_path=drop_path)
        self.p2, self.b2 = WF(ec[1], dc), Block(dc, num_heads, window_size, drop_path=drop_path)
        self.p1 = FeatureRefinementHead(ec[0], dc)
        self.aux_head = AuxHead(dc, n_classes)              # always built: keeps state_dict keys stable
        self.seg_head = nn.Sequential(ConvBNReLU(dc, dc), nn.Dropout2d(dropout), Conv(dc, n_classes, k=1))

    def forward(self, x):
        h, w = x.shape[-2:]
        r1, r2, r3, r4 = self.backbone(x)
        d = self.b4(self.pre_conv(r4))
        aux = self.aux_head(d)
        d = self.b3(self.p3(d, r3))
        d = self.b2(self.p2(d, r2))
        d = self.p1(d, r1)
        out = F.interpolate(self.seg_head(d), (h, w), mode="bilinear", align_corners=False)
        if self.training:
            return out, F.interpolate(aux, (h, w), mode="bilinear", align_corners=False)
        return out

# ---------------------------------------------------------------- EMA
class ModelEMA:
    # Exponential moving average of parameters AND buffers, with a warmup ramp on the decay.
    def __init__(self, model, decay):
        self.decay, self.updates = decay, 0
        self.module = copy.deepcopy(model).eval()
        for p in self.module.parameters(): p.requires_grad_(False)
    @torch.no_grad()
    def update(self, model):
        self.updates += 1
        d = min(self.decay, (1 + self.updates) / (10 + self.updates))
        msd = model.state_dict()
        for k, v in self.module.state_dict().items():
            if v.dtype.is_floating_point: v.mul_(d).add_(msd[k].detach(), alpha=1.0 - d)
            else:                         v.copy_(msd[k])
    def state_dict(self): return {"module": self.module.state_dict(), "updates": self.updates}
    def load_state_dict(self, sd):
        self.module.load_state_dict(sd["module"]); self.updates = sd["updates"]

# ---------------------------------------------------------------- build
def build_model(pretrained=None):
    return UNetFormer(
        encoder=cfg.ENCODER,
        n_classes=cfg.N_CLASSES,
        dc=cfg.DECODE_CHANNELS,
        window_size=cfg.WINDOW_SIZE,
        num_heads=cfg.NUM_HEADS,
        dropout=cfg.DROPOUT,
        drop_path=cfg.DROP_PATH,
        pretrained=cfg.ENCODER_WEIGHTS if pretrained is None else pretrained,
    )

def forward_logits(m, x):
    # eval path: the model already upsamples to the input size and returns a bare tensor
    out = m(x)
    return out[0] if isinstance(out, tuple) else out

model = build_model().cuda()

# Shape smoke test: crash in section 13, not at hour four. Train -> (out, aux), eval -> out.
model.train()
with torch.no_grad():
    _o, _a = model(torch.randn(2, 3, cfg.IMG_SIZE, cfg.IMG_SIZE).cuda())
assert _o.shape == (2, cfg.N_CLASSES, cfg.IMG_SIZE, cfg.IMG_SIZE), _o.shape
assert _a.shape == _o.shape, _a.shape
model.eval()
with torch.no_grad():
    _e = forward_logits(model, torch.randn(1, 3, cfg.IMG_SIZE, cfg.IMG_SIZE).cuda())
assert torch.is_tensor(_e) and _e.shape[1] == cfg.N_CLASSES
del _o, _a, _e; torch.cuda.empty_cache()

# Architecture signature: keys + shapes. Stamped into every checkpoint and verified before any load.
def arch_signature(m):
    s = "|".join(f"{k}:{tuple(v.shape)}" for k, v in sorted(m.state_dict().items()))
    return hashlib.md5(s.encode()).hexdigest()[:12]

ARCH_SIG = arch_signature(model)
ema = ModelEMA(model, cfg.EMA_DECAY) if cfg.EMA_DECAY > 0 else None
eval_model = ema.module if ema is not None else model

enc_p = list(model.backbone.parameters())
enc_ids = {id(p) for p in enc_p}
dec_p = [p for p in model.parameters() if id(p) not in enc_ids]
optimizer = torch.optim.AdamW(
    [{"params": enc_p, "lr": cfg.LR_ENCODER, "name": "encoder"},
     {"params": dec_p, "lr": cfg.LR_DECODER, "name": "decoder"}],
    weight_decay=cfg.WEIGHT_DECAY)
LR_IDX_ENC, LR_IDX_DEC = 0, 1

total_steps = cfg.COS_HORIZON_STEPS          # steps, NOT epochs -- this is the arm-matching line

def lr_lambda(step):
    if step < cfg.WARMUP_STEPS:
        return (step + 1) / cfg.WARMUP_STEPS
    prog = (step - cfg.WARMUP_STEPS) / max(1, total_steps - cfg.WARMUP_STEPS)
    return 0.5 * (1.0 + math.cos(math.pi * min(1.0, prog)))     # clamped: flat zero past the horizon

scheduler = torch.optim.lr_scheduler.LambdaLR(optimizer, lr_lambda)
scaler = torch.amp.GradScaler("cuda", enabled=cfg.AMP)

n_par = sum(p.numel() for p in model.parameters())
print(f"params: {n_par/1e6:.2f} M ({sum(p.numel() for p in enc_p)/1e6:.2f} M encoder / "
      f"{sum(p.numel() for p in dec_p)/1e6:.2f} M decoder)")
print(f"arch signature: {ARCH_SIG} ({cfg.ARCH_VERSION}) | EMA: {'on' if ema else 'off'}")
print(f"steps/pass {steps_per_pass} | cosine horizon {total_steps} steps (identical in both arms) "
      f"| warmup {cfg.WARMUP_STEPS}")
print("LR(dec) at step 0/1500/3000/4500/6000: " + " / ".join(
    f"{lr_lambda(s)*cfg.LR_DECODER:.2e}" for s in (0, 1500, 3000, 4500, 6000)))
```

## 14. Metrics module

Unchanged, and identical to the other Stage-4 notebook — same class, same function, same keys, so
the result JSONs concatenate without any reconciliation. `ConfMat` additionally gets used per base
tile in section 20, which is what the bootstrap resamples.

```python
from skimage.morphology import skeletonize

class ConfMat:
    def __init__(self, n): self.n = n; self.m = torch.zeros(n, n, dtype=torch.int64, device="cuda")
    def update(self, pred, tgt):
        k = tgt != cfg.IGNORE_INDEX
        if k.sum() == 0: return
        self.m += torch.bincount(self.n * tgt[k].long() + pred[k].long(),
                                 minlength=self.n ** 2).reshape(self.n, self.n)
    def numpy(self): return self.m.cpu().numpy().astype(np.float64)

def metrics_from_cm(cm, names):
    tp = np.diag(cm); fp = cm.sum(0) - tp; fn = cm.sum(1) - tp; tot = cm.sum()
    iou = tp / np.maximum(tp + fp + fn, 1e-12)
    f1  = 2 * tp / np.maximum(2 * tp + fp + fn, 1e-12)
    prod = tp / np.maximum(cm.sum(1), 1e-12)
    user = tp / np.maximum(cm.sum(0), 1e-12)
    freq = cm.sum(1) / max(tot, 1e-12)
    po = tp.sum() / max(tot, 1e-12)
    pe = (cm.sum(0) * cm.sum(1)).sum() / max(tot ** 2, 1e-12)
    return {
        "mIoU": float(np.nanmean(iou)), "mF1": float(np.nanmean(f1)),
        "OA": float(po), "FWIoU": float((freq * iou).sum()),
        "kappa": float((po - pe) / max(1 - pe, 1e-12)),
        "per_class": {n: {"IoU": float(iou[i]), "F1": float(f1[i]),
                          "producer_acc": float(prod[i]), "user_acc": float(user[i]),
                          "support_px": int(cm.sum(1)[i])} for i, n in enumerate(names)},
        "confusion_matrix_raw": cm.astype(np.int64).tolist(),
        "confusion_matrix_norm": (cm / np.maximum(cm.sum(1, keepdims=True), 1e-12)).tolist(),
    }

def cldice_pair(pred_bin, gt_bin):
    if gt_bin.sum() == 0 and pred_bin.sum() == 0: return None
    sp = skeletonize(pred_bin.astype(bool)); sg = skeletonize(gt_bin.astype(bool))
    tprec = (sp & gt_bin.astype(bool)).sum() / max(sp.sum(), 1)
    tsens = (sg & pred_bin.astype(bool)).sum() / max(sg.sum(), 1)
    if tprec + tsens == 0: return 0.0
    return float(2 * tprec * tsens / (tprec + tsens))
```

## 15. Efficiency metrics

Both arms run the same architecture, so these are identical by construction — they are recorded so
the Stage-4 table can stand alone rather than pointing back at part 1.

The `FlopCounterMode` cross-check is kept from part 1 and matters for this model specifically:
`ptflops` counts modules, and the windowed attention's `q @ k` and `attn @ v` are bare tensor ops,
so the module-based number understates UNetFormer. Report both columns or report the traced one.

```python
def efficiency_report(m):
    from ptflops import get_model_complexity_info
    m.eval()
    with torch.amp.autocast("cuda", enabled=False):
        macs, params = get_model_complexity_info(
            m, (3, cfg.IMG_SIZE, cfg.IMG_SIZE),
            as_strings=False, print_per_layer_stat=False, verbose=False)
    x = torch.randn(1, 3, cfg.IMG_SIZE, cfg.IMG_SIZE, device="cuda")
    # Cross-check: ptflops misses the attention matmuls (tensor ops, not modules).
    try:
        from torch.utils.flop_counter import FlopCounterMode
        fc = FlopCounterMode(display=False)
        with fc, torch.no_grad():
            m(x)
        traced_gflops = fc.get_total_flops() / 1e9
    except Exception as e:
        print("flop_counter unavailable:", type(e).__name__); traced_gflops = float("nan")
    with torch.no_grad(), torch.amp.autocast("cuda", enabled=cfg.AMP):
        for _ in range(20): m(x)
        torch.cuda.synchronize(); t0 = time.time()
        for _ in range(50): m(x)
        torch.cuda.synchronize()
    return {"params_M": params / 1e6,
            "GFLOPs_512": macs * 2 / 1e9,           # ptflops returns MACs; FLOPs = 2 x MACs
            "GFLOPs_512_flopcounter": traced_gflops,
            "inference_fps_bs1": 50 / (time.time() - t0)}
```

## 18. Train / validate loops

`batch_stream()` is an endless iterator over the train loader. It rebuilds the loader's iterator
whenever it runs dry and counts passes for logging, which is the only bookkeeping the step-based
loop needs — augmentation no longer depends on knowing which pass it is on.

Two UNetFormer-specific details:

- The model returns `(out, aux)` in train mode and a bare tensor in eval, so `train_steps` unpacks
  and adds `AUX_WEIGHT * criterion(aux, y)[0]`. `scaler.unscale_` runs before `clip_grad_norm_`,
  otherwise the 1.0 threshold would be applied to loss-scaled gradients and mean nothing.
- The EMA is updated only when the scaler did not skip the step. An AMP overflow leaves the weights
  untouched, and folding an unchanged model into the EMA would silently slow its ramp.

Validation is the full 408 real val images every time, on the **EMA** weights — those are the
reported numbers. The raw-weight pass is a diagnostic only (`VAL_RAW_DIAG`); when EMA and raw
diverge sharply it usually means the decay is too slow for the horizon, which is worth knowing
before the paper quotes the EMA number.

```python
PASS_COUNT = {"n": 0}

def batch_stream():
    while True:
        for b in train_ld:
            yield b
        PASS_COUNT["n"] += 1

@torch.no_grad()
def validate(m=None):
    m = m if m is not None else eval_model
    m.eval(); cm = ConfMat(cfg.N_CLASSES); tot, n = 0.0, 0
    for x, y in val_ld:
        x = x.cuda(non_blocking=True); y = y.cuda(non_blocking=True)
        with torch.amp.autocast("cuda", enabled=cfg.AMP):
            logits = forward_logits(m, x)
        loss, _, _ = criterion(logits, y)
        cm.update(logits.float().argmax(1), y)
        tot += float(loss) * x.size(0); n += x.size(0)
    return tot / max(n, 1), metrics_from_cm(cm.numpy(), cfg.CLASS_NAMES)

def train_steps(stream, n_steps, step0):
    model.train()
    tot, seen, gn_acc, aux_acc = 0.0, 0, [], []
    for k in range(n_steps):
        x, y = next(stream)
        x = x.cuda(non_blocking=True); y = y.cuda(non_blocking=True)
        with torch.amp.autocast("cuda", enabled=cfg.AMP):
            out = model(x)
        logits, aux = out if isinstance(out, tuple) else (out, None)
        loss, ce, lv = criterion(logits, y)                             # fp32, outside autocast
        if aux is not None and cfg.AUX_WEIGHT > 0:
            la = criterion(aux, y)[0]                                   # deep supervision
            loss = loss + cfg.AUX_WEIGHT * la
            aux_acc.append(float(la))
        scaler.scale(loss / cfg.GRAD_ACCUM).backward()

        if (k + 1) % cfg.GRAD_ACCUM == 0:
            scaler.unscale_(optimizer)                                  # unscale BEFORE clipping
            gn = torch.nn.utils.clip_grad_norm_(model.parameters(), cfg.CLIP_NORM)
            gn_acc.append(float(gn))
            prev_scale = scaler.get_scale()
            scaler.step(optimizer); scaler.update()
            optimizer.zero_grad(set_to_none=True); scheduler.step()
            if ema is not None and scaler.get_scale() >= prev_scale:
                ema.update(model)

        tot += float(loss) * x.size(0); seen += x.size(0)
        if k % 50 == 0:
            print(f"  s{step0+k} loss {float(loss):.4f} (ce {float(ce):.4f} lov {float(lv):.4f}"
                  f"{f' aux {aux_acc[-1]:.4f}' if aux_acc else ''}) "
                  f"gnorm {gn_acc[-1] if gn_acc else 0:.3f}", flush=True)
    return (tot / max(seen, 1), float(np.mean(gn_acc)), float(np.max(gn_acc)),
            float(np.mean(aux_acc)) if aux_acc else float("nan"))
```

## 19. Main loop with automatic resume

On start this pulls `last.pt` from the arm's repo and verifies `arch_sig`, `arch_version` and
`ARM` against what section 13 built. A mismatch on the architecture discards the stale checkpoints
and starts clean rather than half-loading; a mismatch on the arm is an assert, because that means
the wrong repo for this config and nothing good follows from continuing.

Early stopping is on val mIoU with a patience of 8 validations. `MAX_HOURS` stops the session
cleanly and pushes `last.pt`; re-running the notebook with the same `ARM` picks up from that step.

```python
step, best_miou, best_step, bad, history = 0, -1.0, -1, 0, []
best_metrics = {}

ck = hf_try_download("last.pt") if cfg.RESUME else None
if ck:
    st = torch.load(ck, map_location="cuda", weights_only=False)
    assert st["cfg"]["ARM"] == cfg.ARM, (
        f"last.pt is arm {st['cfg']['ARM']} but this session is {cfg.ARM} -- wrong repo")
    same_arch = (st.get("arch_sig") == ARCH_SIG and st.get("arch_version") == cfg.ARCH_VERSION)
    if not same_arch:
        print(f"ARCHITECTURE MISMATCH: checkpoint {st.get('arch_version')} / {st.get('arch_sig')} "
              f"vs current {cfg.ARCH_VERSION} / {ARCH_SIG}")
        print("  -> discarding stale checkpoints and starting fresh")
        hf_delete("last.pt"); hf_delete("best.pt")
    else:
        model.load_state_dict(st["model_state"])
        if ema is not None and st.get("ema_state") is not None: ema.load_state_dict(st["ema_state"])
        optimizer.load_state_dict(st["optim_state"])
        scheduler.load_state_dict(st["sched_state"]); scaler.load_state_dict(st["scaler_state"])
        step = st["step"]; best_miou = st["best_miou"]; best_step = st["best_step"]
        bad = st["bad"]; history = st["history"]; best_metrics = st.get("best_metrics", {})
        print(f"RESUMED at step {step} (best mIoU {best_miou:.4f} @ s{best_step}, bad={bad})")
else:
    print("fresh start")

torch.cuda.reset_peak_memory_stats()
t_start, stream = time.time(), batch_stream()
n_val = 0
STOP = None

while step < cfg.MAX_STEPS:
    n = min(cfg.VAL_EVERY_STEPS, cfg.MAX_STEPS - step)
    t0 = time.time()
    tr_loss, gn_mean, gn_max, aux_mean = train_steps(stream, n, step)
    step += n; n_val += 1
    va_loss, m = validate()                                   # EMA -- the reported numbers
    m_raw = validate(model)[1] if cfg.VAL_RAW_DIAG else m      # raw -- diagnostic only
    peak_gb = torch.cuda.max_memory_allocated() / 1e9
    row = {"step": step, "pass": PASS_COUNT["n"], "train_loss": tr_loss, "train_aux": aux_mean,
           "val_loss": va_loss, "overfit_gap": va_loss - tr_loss,
           "val_mIoU": m["mIoU"], "val_mIoU_raw": m_raw["mIoU"], "val_mF1": m["mF1"],
           "val_OA": m["OA"], "val_kappa": m["kappa"], "grad_norm_mean": gn_mean,
           "grad_norm_max": gn_max, "lr_enc": scheduler.get_last_lr()[LR_IDX_ENC],
           "lr_dec": scheduler.get_last_lr()[LR_IDX_DEC], "peak_vram_gb": peak_gb,
           "block_sec": time.time() - t0,
           **{f"IoU/{k}": v["IoU"] for k, v in m["per_class"].items()}}
    history.append(row)
    if WANDB: wandb.log(row, step=step)
    print(f"[s{step}] train {tr_loss:.4f} | val {va_loss:.4f} | mIoU(ema) {m['mIoU']:.4f} "
          f"| mIoU(raw) {m_raw['mIoU']:.4f} | River {m['per_class']['River']['IoU']:.4f} "
          f"Road {m['per_class']['Road']['IoU']:.4f} | {row['block_sec']:.0f}s "
          f"| vram {peak_gb:.1f}GB", flush=True)

    if m["mIoU"] > best_miou:
        best_miou, best_step, bad, best_metrics = m["mIoU"], step, 0, m
        p = f"{cfg.OUT_DIR}/best.pt"
        torch.save({"model_state": eval_model.state_dict(), "cfg": asdict(cfg),
                    "metrics": m, "step": step, "arch_sig": ARCH_SIG,
                    "arch_version": cfg.ARCH_VERSION,
                    "weights": "ema" if ema is not None else "raw"}, p)
        hf_push(p, "best.pt", f"best mIoU {best_miou:.4f} @ s{step}")
        push_model_card(best_metrics, step)
        print(f"  -> new best {best_miou:.4f}, pushed")
    else:
        bad += 1

    elapsed_h = (time.time() - t_start) / 3600
    if bad >= cfg.PATIENCE_VALS: STOP = "early stop (patience)"
    elif elapsed_h > cfg.MAX_HOURS: STOP = "session budget"
    elif step >= cfg.MAX_STEPS: STOP = "max steps"

    if (n_val % cfg.PUSH_LAST_EVERY_VALS == 0) or STOP:
        p = f"{cfg.OUT_DIR}/last.pt"
        torch.save({"model_state": model.state_dict(), "optim_state": optimizer.state_dict(),
                    "sched_state": scheduler.state_dict(), "scaler_state": scaler.state_dict(),
                    "ema_state": (ema.state_dict() if ema is not None else None),
                    "cfg": asdict(cfg), "arch_sig": ARCH_SIG, "arch_version": cfg.ARCH_VERSION,
                    "step": step, "best_miou": best_miou, "best_step": best_step,
                    "bad": bad, "best_metrics": best_metrics, "history": history}, p)
        hf_push(p, "last.pt", f"state s{step}")

    if STOP:
        print(f"STOP: {STOP} at step {step} (best {best_miou:.4f} @ s{best_step})")
        break

wall_h = (time.time() - t_start) / 3600
print(f"session done | best mIoU {best_miou:.4f} @ step {best_step} | {wall_h:.2f} h")
if STOP == "session budget":
    print("RE-RUN THIS NOTEBOOK WITH THE SAME ARM -- it resumes from last.pt")
```

## 20. Final evaluation, per tile

Two things happen here that part 1 did not do.

**Per-base-tile confusion matrices.** The 136 val tiles each get their own 6x6 matrix, summed over
their three flip variants. Summing all 136 reproduces the pooled result exactly, so nothing is lost
— but keeping them separate is what makes the next part possible.

**A bootstrap confidence interval.** Resampling the 136 tiles with replacement 2000 times gives the
spread of mIoU attributable to *which tiles happened to land in this fold's validation block*. It
costs nothing, and with one seed per arm it is the only honest uncertainty estimate available. The
tile matrices are saved so section 22 can bootstrap the R1 - R0 **difference** on paired resamples,
which is a much tighter statistic than comparing two independent intervals.

`best.pt` holds EMA weights and its `arch_sig` is asserted against the live model before loading.

```python
_bp = f"{cfg.OUT_DIR}/best.pt"
if not os.path.exists(_bp):
    # session resumed past the end, or the loop never improved locally -- pull the best from the Hub
    _bp = hf_try_download("best.pt")
    assert _bp, "no best.pt locally or on the Hub -- training never completed a validation"
    print("best.pt pulled from the Hub")
_ck = torch.load(_bp, map_location="cuda", weights_only=False)
assert _ck.get("arch_sig") == ARCH_SIG, (
    f"best.pt architecture {_ck.get('arch_sig')} != current {ARCH_SIG}")
best_model = build_model(pretrained=False).cuda()
best_model.load_state_dict(_ck["model_state"])
best_model.eval()
print(f"evaluating best.pt from step {_ck['step']} ({_ck.get('weights','raw')} weights), "
      f"reported mIoU {_ck['metrics']['mIoU']:.4f}")

VAL_TILE_IDS = sorted(va_ids)
tile_pos = {b: i for i, b in enumerate(VAL_TILE_IDS)}
TILE_CM = np.zeros((len(VAL_TILE_IDS), cfg.N_CLASSES, cfg.N_CLASSES), dtype=np.int64)
cld = {1: [], 2: []}                                   # Road, River

with torch.no_grad():
    for j in range(0, len(va_recs), cfg.BATCH_SIZE):
        chunk = va_recs[j:j + cfg.BATCH_SIZE]
        xs, ys = zip(*[val_ds[j + k] for k in range(len(chunk))])
        x = torch.stack(xs).cuda(); y = torch.stack(ys).cuda()
        with torch.amp.autocast("cuda", enabled=cfg.AMP):
            logits = forward_logits(best_model, x)
        pred = logits.float().argmax(1)
        for b, rec in enumerate(chunk):
            c = ConfMat(cfg.N_CLASSES); c.update(pred[b], y[b])
            TILE_CM[tile_pos[rec[0]]] += c.numpy().astype(np.int64)
        pn, yn = pred.cpu().numpy(), y.cpu().numpy()
        valid = yn != cfg.IGNORE_INDEX
        for b in range(pn.shape[0]):
            for c_ in cld:
                v = cldice_pair((pn[b] == c_) & valid[b], (yn[b] == c_) & valid[b])
                if v is not None: cld[c_].append(v)

CM_TOTAL = TILE_CM.sum(0).astype(np.float64)
final = metrics_from_cm(CM_TOTAL, cfg.CLASS_NAMES)
final["clDice"] = {cfg.CLASS_NAMES[c]: (float(np.mean(v)) if v else None) for c, v in cld.items()}
final["clDice_n_images"] = {cfg.CLASS_NAMES[c]: len(v) for c, v in cld.items()}

def boot_metrics(tile_cm, n_boot, seed=0):
    rs = np.random.RandomState(seed)
    N = tile_cm.shape[0]
    mious, per_cls = [], []
    for _ in range(n_boot):
        pick = rs.randint(0, N, N)
        cm = tile_cm[pick].sum(0).astype(np.float64)
        tp = np.diag(cm); fp = cm.sum(0) - tp; fn = cm.sum(1) - tp
        den = tp + fp + fn
        iou = np.where(den > 0, tp / np.maximum(den, 1e-12), np.nan)   # absent class -> excluded
        mious.append(float(np.nanmean(iou))); per_cls.append(iou)
    return np.array(mious), np.array(per_cls)

bm, bpc = boot_metrics(TILE_CM, cfg.BOOTSTRAP_N, cfg.SEED)
final["bootstrap"] = {
    "n": cfg.BOOTSTRAP_N, "unit": "val base tile (136)",
    "mIoU_mean": float(bm.mean()), "mIoU_ci95": [float(np.percentile(bm, 2.5)),
                                                 float(np.percentile(bm, 97.5))],
    "per_class_ci95": {n: [float(np.percentile(bpc[:, i], 2.5)),
                           float(np.percentile(bpc[:, i], 97.5))]
                       for i, n in enumerate(cfg.CLASS_NAMES)}}

final.update(efficiency_report(best_model))
final["peak_train_vram_gb"] = float(torch.cuda.max_memory_allocated() / 1e9)
final["train_wall_clock_h_this_session"] = wall_h
final["best_step"] = best_step
final["run"] = {"model": MODEL_TAG, "arm": cfg.ARM, "split_mode": cfg.SPLIT_MODE,
                "fold": cfg.FOLD, "seed": cfg.SEED, "n_train": len(tr_recs),
                "n_real": len(real_tr_recs), "n_fake": len(fake_tr_recs),
                "train_sha1": split_sha1, "weights_from": cfg.WEIGHTS_FROM,
                "arch_version": cfg.ARCH_VERSION, "arch_sig": ARCH_SIG,
                "weights": _ck.get("weights", "raw"), "aux_weight": cfg.AUX_WEIGHT,
                "lr_decoder": cfg.LR_DECODER, "ema_decay": cfg.EMA_DECAY}

print(f"arm {cfg.ARM} | mIoU {final['mIoU']:.4f} "
      f"[{final['bootstrap']['mIoU_ci95'][0]:.4f}, {final['bootstrap']['mIoU_ci95'][1]:.4f}] "
      f"| mF1 {final['mF1']:.4f} | OA {final['OA']:.4f} | kappa {final['kappa']:.4f}")
print("\nper class:")
for i, (k, v) in enumerate(final["per_class"].items()):
    lo, hi = final["bootstrap"]["per_class_ci95"][k]
    print(f"  {k:<14} IoU {v['IoU']:.4f} [{lo:.4f}, {hi:.4f}]  F1 {v['F1']:.4f}")
print("clDice:", {k: (round(v, 4) if v is not None else None)
                  for k, v in final["clDice"].items()})
```

## 22. R0 vs R1 — the actual Stage-4 result

**Run this only after both arms have finished.** It pulls both result sets from the Hub and
computes the comparison the paper reports.

The headline number is a **paired bootstrap**: each of the 2000 resamples draws the same 136 tile
indices for both arms and takes the difference, so tile-to-tile difficulty cancels out instead of
adding two independent variances together. The reported p-value is the fraction of resamples in
which the difference has the opposite sign to the observed one — a two-sided sign test over the
resample distribution, not a t-test, and it makes no normality assumption.

One thing this cannot tell you: both arms ran a single seed, so the interval covers sampling
variation across validation tiles, not variation across training runs. State that in Limitations
rather than letting the interval imply more than it measures.

```python
import pandas as pd

def pull_arm(arm):
    repo = f"{cfg.HF_USER}/{cfg.HF_PREFIX}-{arm}-{cfg.SPLIT_MODE}-fold{cfg.FOLD}"
    t = f"{arm}_{cfg.SPLIT_MODE}_fold{cfg.FOLD}"
    res = json.load(open(hf_hub_download(repo, f"results_{t}.json", token=HF_TOKEN)))
    cm  = np.load(hf_hub_download(repo, f"tile_cm_{arm}.npy", token=HF_TOKEN))
    idx = json.load(open(hf_hub_download(repo, f"tile_index_{arm}.json", token=HF_TOKEN)))
    return res, cm, idx

try:
    R0, CM0, IX0 = pull_arm("R0")
    R1, CM1, IX1 = pull_arm("R1")
except Exception as e:
    print(f"both arms not finished yet ({type(e).__name__}) -- run this after the second one")
    R0 = None

if R0 is not None:
    assert IX0["val_tile_ids"] == IX1["val_tile_ids"], "arms validated on different tiles"
    assert IX0["train_sha1"] == IX1["train_sha1"], "arms used different real splits"
    N = CM0.shape[0]

    def iou_from(cm):
        tp = np.diag(cm); fp = cm.sum(0) - tp; fn = cm.sum(1) - tp
        den = tp + fp + fn
        return np.where(den > 0, tp / np.maximum(den, 1e-12), np.nan)

    rs = np.random.RandomState(cfg.SEED)
    d_miou, d_cls = [], []
    for _ in range(cfg.BOOTSTRAP_N):
        pick = rs.randint(0, N, N)                       # SAME tiles for both arms
        i0 = iou_from(CM0[pick].sum(0).astype(np.float64))
        i1 = iou_from(CM1[pick].sum(0).astype(np.float64))
        d_miou.append(np.nanmean(i1) - np.nanmean(i0)); d_cls.append(i1 - i0)
    # per-class deltas can carry nan if a class vanished from a resample; nan-aware percentiles below
    d_miou = np.array(d_miou); d_cls = np.array(d_cls)

    obs = R1["mIoU"] - R0["mIoU"]
    p = min(1.0, float(np.mean(d_miou <= 0) if obs > 0 else np.mean(d_miou >= 0)) * 2)
    print(f"{MODEL_TAG}  {cfg.SPLIT_MODE} fold {cfg.FOLD}\n")
    print(f"{'metric':<16}{'R0':>10}{'R1':>10}{'delta':>10}{'95% CI on delta':>24}")
    print(f"{'mIoU':<16}{R0['mIoU']:10.4f}{R1['mIoU']:10.4f}{obs:+10.4f}"
          f"   [{np.percentile(d_miou,2.5):+.4f}, {np.percentile(d_miou,97.5):+.4f}]")
    for k in ["mF1", "OA", "FWIoU", "kappa"]:
        print(f"{k:<16}{R0[k]:10.4f}{R1[k]:10.4f}{R1[k]-R0[k]:+10.4f}")
    print(f"\npaired-bootstrap two-sided p on mIoU: {p:.4f}  (n={cfg.BOOTSTRAP_N} resamples)")

    print(f"\n{'class':<14}{'IoU R0':>9}{'IoU R1':>9}{'delta':>9}{'95% CI on delta':>24}")
    rows = []
    for i, nm in enumerate(cfg.CLASS_NAMES):
        a, b = R0["per_class"][nm]["IoU"], R1["per_class"][nm]["IoU"]
        lo, hi = np.nanpercentile(d_cls[:, i], 2.5), np.nanpercentile(d_cls[:, i], 97.5)
        star = " *" if (lo > 0 or hi < 0) else ""
        print(f"{nm:<14}{a:9.4f}{b:9.4f}{b-a:+9.4f}   [{lo:+.4f}, {hi:+.4f}]{star}")
        rows.append({"class": nm, "IoU_R0": a, "IoU_R1": b, "delta": b - a,
                     "ci_lo": lo, "ci_hi": hi})
    print("\n* = 95% CI on the difference excludes zero")

    print(f"\n{'clDice':<14}{'R0':>9}{'R1':>9}{'delta':>9}")
    for k in R0["clDice"]:
        a, b = R0["clDice"][k], R1["clDice"][k]
        if a is not None and b is not None:
            print(f"{k:<14}{a:9.4f}{b:9.4f}{b-a:+9.4f}")

    comp = {"model": MODEL_TAG,
            "mIoU_R0": R0["mIoU"], "mIoU_R1": R1["mIoU"], "delta_mIoU": obs,
            "delta_mIoU_ci95": [float(np.percentile(d_miou, 2.5)),
                                float(np.percentile(d_miou, 97.5))],
            "p_two_sided": p, "n_boot": cfg.BOOTSTRAP_N, "per_class": rows,
            "best_step": {"R0": R0["best_step"], "R1": R1["best_step"]}}
    json.dump(comp, open(f"{cfg.OUT_DIR}/stage4_R0_vs_R1.json", "w"), indent=2, default=float)
    pd.DataFrame(rows).to_csv(f"{cfg.OUT_DIR}/stage4_R0_vs_R1.csv", index=False)
    hf_push(f"{cfg.OUT_DIR}/stage4_R0_vs_R1.json", "stage4_R0_vs_R1.json", "arm comparison")
    hf_push(f"{cfg.OUT_DIR}/stage4_R0_vs_R1.csv", "stage4_R0_vs_R1.csv", "arm comparison")
```

## 23. Standalone inference from Hugging Face

Self-contained: paste into any notebook with `timm`, `einops` and a token, and it rebuilds the
model from the checkpoint's own embedded `cfg` rather than from this notebook's. `best.pt` holds
the **EMA** weights. Nothing above this cell is required, which is why the architecture is
redefined here rather than reused — that is deliberate duplication, not an oversight.

```python
# ============ STANDALONE: everything needed is inside this cell ============
import os, numpy as np, cv2, torch, torch.nn as nn, torch.nn.functional as F, timm
from einops import rearrange
from huggingface_hub import hf_hub_download

class Conv(nn.Sequential):
    def __init__(self, i, o, k=3, stride=1, dilation=1, bias=False):
        super().__init__(nn.Conv2d(i, o, k, stride, ((stride-1)+dilation*(k-1))//2, dilation, bias=bias))

class ConvBN(nn.Sequential):
    def __init__(self, i, o, k=3, stride=1, dilation=1):
        super().__init__(nn.Conv2d(i, o, k, stride, ((stride-1)+dilation*(k-1))//2, dilation, bias=False),
                         nn.BatchNorm2d(o))

class ConvBNReLU(nn.Sequential):
    def __init__(self, i, o, k=3, stride=1, dilation=1):
        super().__init__(nn.Conv2d(i, o, k, stride, ((stride-1)+dilation*(k-1))//2, dilation, bias=False),
                         nn.BatchNorm2d(o), nn.ReLU6(inplace=True))

class SeparableConvBN(nn.Sequential):
    def __init__(self, i, o, k=3, stride=1, dilation=1):
        super().__init__(nn.Conv2d(i, i, k, stride, ((stride-1)+dilation*(k-1))//2, dilation,
                                   groups=i, bias=False),
                         nn.Conv2d(i, o, 1, bias=False), nn.BatchNorm2d(o))

class DropPath(nn.Module):
    def __init__(self, p=0.0):
        super().__init__(); self.p = float(p)
    def forward(self, x):
        if self.p == 0.0 or not self.training: return x
        keep = 1.0 - self.p
        return x * x.new_empty((x.shape[0],) + (1,) * (x.ndim - 1)).bernoulli_(keep) / keep

class Mlp(nn.Module):
    def __init__(self, dim, hidden, drop=0.0):
        super().__init__()
        self.fc1, self.act = nn.Conv2d(dim, hidden, 1), nn.ReLU6(inplace=True)
        self.fc2, self.drop = nn.Conv2d(hidden, dim, 1), nn.Dropout(drop)
    def forward(self, x): return self.drop(self.fc2(self.drop(self.act(self.fc1(x)))))

class GlobalLocalAttention(nn.Module):
    def __init__(self, dim=64, num_heads=8, window_size=8, qkv_bias=False):
        super().__init__()
        self.nh, self.ws = num_heads, window_size
        self.scale = (dim // num_heads) ** -0.5
        self.qkv = Conv(dim, 3 * dim, k=1, bias=qkv_bias)
        self.local1, self.local2 = ConvBN(dim, dim, k=3), ConvBN(dim, dim, k=1)
        self.proj = SeparableConvBN(dim, dim, k=window_size)
        self.attn_x = nn.AvgPool2d((window_size, 1), stride=1, padding=(window_size // 2 - 1, 0))
        self.attn_y = nn.AvgPool2d((1, window_size), stride=1, padding=(0, window_size // 2 - 1))
        ws = window_size
        self.rpb_table = nn.Parameter(torch.zeros((2 * ws - 1) ** 2, num_heads))
        c = torch.stack(torch.meshgrid(torch.arange(ws), torch.arange(ws), indexing="ij")).flatten(1)
        rel = (c[:, :, None] - c[:, None, :]).permute(1, 2, 0).contiguous()
        rel[:, :, 0] += ws - 1; rel[:, :, 1] += ws - 1; rel[:, :, 0] *= 2 * ws - 1
        self.register_buffer("rpi", rel.sum(-1), persistent=False)
    def forward(self, x):
        B, C, H, W = x.shape
        local = self.local2(x) + self.local1(x)
        ws = self.ws
        ph, pw = (ws - H % ws) % ws, (ws - W % ws) % ws
        xp = F.pad(x, (0, pw, 0, ph), mode="reflect") if (ph or pw) else x
        Hp, Wp = xp.shape[-2:]
        q, k, v = rearrange(self.qkv(xp), "b (qkv h d) (hh w1) (ww w2) -> qkv (b hh ww) h (w1 w2) d",
                            qkv=3, h=self.nh, w1=ws, w2=ws)
        bias = self.rpb_table[self.rpi.view(-1)].view(ws * ws, ws * ws, self.nh).permute(2, 0, 1)
        attn = ((q @ k.transpose(-2, -1)) * self.scale + bias.unsqueeze(0)).softmax(-1) @ v
        attn = rearrange(attn, "(b hh ww) h (w1 w2) d -> b (h d) (hh w1) (ww w2)",
                         b=B, hh=Hp // ws, ww=Wp // ws, w1=ws, w2=ws)[:, :, :H, :W]
        fused = attn + local
        cross = (self.attn_x(F.pad(fused, (0, 0, 0, 1), mode="reflect")) +
                 self.attn_y(F.pad(fused, (0, 1, 0, 0), mode="reflect")))
        return self.proj(F.pad(fused + cross, (0, 1, 0, 1), mode="reflect"))[:, :, :H, :W]

class Block(nn.Module):
    def __init__(self, dim=64, num_heads=8, window_size=8, mlp_ratio=4.0, drop=0.0, drop_path=0.0):
        super().__init__()
        self.norm1, self.norm2 = nn.BatchNorm2d(dim), nn.BatchNorm2d(dim)
        self.attn = GlobalLocalAttention(dim, num_heads, window_size)
        self.mlp = Mlp(dim, int(dim * mlp_ratio), drop)
        self.dp1, self.dp2 = DropPath(drop_path), DropPath(drop_path)
    def forward(self, x):
        x = x + self.dp1(self.attn(self.norm1(x)))
        return x + self.dp2(self.mlp(self.norm2(x)))

class WF(nn.Module):
    def __init__(self, in_ch, dc, eps=1e-8):
        super().__init__()
        self.pre_conv, self.post_conv = Conv(in_ch, dc, k=1), ConvBNReLU(dc, dc, k=3)
        self.weights, self.eps = nn.Parameter(torch.ones(2, dtype=torch.float32)), eps
    def forward(self, x, res):
        x = F.interpolate(x, size=res.shape[-2:], mode="bilinear", align_corners=False)
        w = torch.relu(self.weights); w = w / (w.sum() + self.eps)
        return self.post_conv(w[0] * self.pre_conv(res) + w[1] * x)

class FeatureRefinementHead(nn.Module):
    def __init__(self, in_ch, dc, eps=1e-8):
        super().__init__()
        self.pre_conv, self.post_conv = Conv(in_ch, dc, k=1), ConvBNReLU(dc, dc, k=3)
        self.weights, self.eps = nn.Parameter(torch.ones(2, dtype=torch.float32)), eps
        self.pa = nn.Sequential(nn.Conv2d(dc, dc, 3, padding=1, groups=dc), nn.Sigmoid())
        self.ca = nn.Sequential(nn.AdaptiveAvgPool2d(1), Conv(dc, max(dc // 16, 4), k=1), nn.ReLU6(True),
                                Conv(max(dc // 16, 4), dc, k=1), nn.Sigmoid())
        self.shortcut, self.proj = ConvBN(dc, dc, k=1), SeparableConvBN(dc, dc, k=3)
        self.act = nn.ReLU6(True)
    def forward(self, x, res):
        x = F.interpolate(x, size=res.shape[-2:], mode="bilinear", align_corners=False)
        w = torch.relu(self.weights); w = w / (w.sum() + self.eps)
        x = self.post_conv(w[0] * self.pre_conv(res) + w[1] * x)
        sc = self.shortcut(x)
        x = self.pa(x) * x + self.ca(x) * x
        return self.act(self.proj(x) + sc)

class AuxHead(nn.Module):
    def __init__(self, dc, n_cls):
        super().__init__()
        self.body = nn.Sequential(ConvBNReLU(dc, dc), nn.Dropout2d(0.1), Conv(dc, n_cls, k=1))
    def forward(self, x): return self.body(x)

class UNetFormer(nn.Module):
    def __init__(self, encoder="resnet18", n_classes=6, dc=64, window_size=8,
                 num_heads=8, dropout=0.1, drop_path=0.1, pretrained=True):
        super().__init__()
        self.backbone = timm.create_model(encoder, features_only=True,
                                          out_indices=(1, 2, 3, 4), pretrained=pretrained)
        ec = self.backbone.feature_info.channels()
        self.pre_conv = ConvBN(ec[3], dc, k=1)
        self.b4 = Block(dc, num_heads, window_size, drop_path=drop_path)
        self.p3, self.b3 = WF(ec[2], dc), Block(dc, num_heads, window_size, drop_path=drop_path)
        self.p2, self.b2 = WF(ec[1], dc), Block(dc, num_heads, window_size, drop_path=drop_path)
        self.p1 = FeatureRefinementHead(ec[0], dc)
        self.aux_head = AuxHead(dc, n_classes)
        self.seg_head = nn.Sequential(ConvBNReLU(dc, dc), nn.Dropout2d(dropout), Conv(dc, n_classes, k=1))
    def forward(self, x):
        h, w = x.shape[-2:]
        r1, r2, r3, r4 = self.backbone(x)
        d = self.b4(self.pre_conv(r4))
        aux = self.aux_head(d)
        d = self.b3(self.p3(d, r3))
        d = self.b2(self.p2(d, r2))
        d = self.p1(d, r1)
        out = F.interpolate(self.seg_head(d), (h, w), mode="bilinear", align_corners=False)
        if self.training:
            return out, F.interpolate(aux, (h, w), mode="bilinear", align_corners=False)
        return out

def load_from_hf(repo_id, token=None, device="cuda"):
    ck = torch.load(hf_hub_download(repo_id, "best.pt", token=token),
                    map_location=device, weights_only=False)
    c = ck["cfg"]
    m = UNetFormer(encoder=c["ENCODER"], n_classes=c["N_CLASSES"], dc=c["DECODE_CHANNELS"],
                   window_size=c["WINDOW_SIZE"], num_heads=c["NUM_HEADS"], dropout=c["DROPOUT"],
                   drop_path=c.get("DROP_PATH", 0.0),
                   pretrained=False).to(device).eval()      # weights come from the checkpoint
    m.load_state_dict(ck["model_state"])
    return m, c, ck["metrics"]

@torch.no_grad()
def predict_paths(model, c, paths, device="cuda", batch=8):
    mean = np.array([0.485, 0.456, 0.406], np.float32); std = np.array([0.229, 0.224, 0.225], np.float32)
    out = []
    for i in range(0, len(paths), batch):
        xs = []
        for p in paths[i:i + batch]:
            im = cv2.imread(str(p), cv2.IMREAD_COLOR)
            im = cv2.resize(im, (c["IMG_SIZE"], c["IMG_SIZE"]), interpolation=cv2.INTER_AREA)
            a = cv2.cvtColor(im, cv2.COLOR_BGR2RGB).astype(np.float32) / 255.0
            xs.append(torch.from_numpy(((a - mean) / std).transpose(2, 0, 1)))
        x = torch.stack(xs).to(device)
        with torch.amp.autocast("cuda"):
            out.append(model(x).float().argmax(1).cpu().numpy())
    return np.concatenate(out)

# --- example ---
inf_model, inf_cfg, inf_metrics = load_from_hf(HF_REPO, token=HF_TOKEN)
print(f"loaded {HF_REPO} | arm {inf_cfg['ARM']} | val mIoU {inf_metrics['mIoU']:.4f}")
demo = [IMG_DIR / r[2] for r in va_recs[:4]]
preds = predict_paths(inf_model, inf_cfg, demo)
print("pred shape:", preds.shape, "| classes present:", np.unique(preds).tolist())
```

### Asserts that must pass before you walk away
- Section 5: `channel-order check PASSED`
- Section 8: `four-way agreement with S1/S2/S3 ASSERTED` — if this fails, the real split
  disagrees with what the generators were trained on and the whole experiment is void
- Section 9 (R1 only): 804 synthetic pairs loaded, no synthetic record in validation
- Section 11: `augmentation liveness check PASSED`
- Section 13: shape smoke test passes and `arch signature` matches the one in any `last.pt` you
  are resuming from (section 19 enforces this and clears the checkpoint if it does not)

### Cross-model consistency
`UPSTREAM_TRAIN_SHA1`, `MAX_STEPS`, `WARMUP_STEPS`, `VAL_EVERY_STEPS`, `PATIENCE_VALS`,
`WEIGHTS_FROM`, `AUG_FLIP_FAKE`, `BOOTSTRAP_N` and `SEED` are all identical to
`geonusaf-stage4-segformer-b0-R0R1.ipynb`. If you change any of them here, change them there too
and re-run both models, or the Stage-4 table stops being a like-for-like comparison. The one thing
that legitimately differs is wall-clock: UNetFormer is roughly 3x the GFLOPs of SegFormer-B0 at
512, so budget 3-5 h per arm rather than 2-3.

### Reading the result
Section 22's mIoU delta is the Stage-4 headline for this architecture, and the per-class table is
where the actual argument lives. River and Road are the classes the synthetic data was supposed to
help; a `*` next to them means the 95% interval on the difference excludes zero.

**A negative delta is a publishable result.** "Synthetic pairs at 50% mixing did not improve
fold-1 segmentation, and here is the layout-fidelity and memorization evidence for why" is a real
contribution given Stage 3's numbers are already measured. Do not tune the mixing ratio against
the validation set to find a positive number — that is selection on the test set, and with one
fold and one seed it would be indefensible.

If the two architectures disagree in sign, that is itself the finding, and it is a stronger one
than either arm pair alone: it says the effect of synthetic data here is model-dependent rather
than a property of the data. Report both, do not pick the flattering one.

### What this run cannot support
One seed, one fold, one mixing ratio. Every claim has to be scoped to those three facts in the
paper, plus the architecture. The bootstrap interval covers tile sampling only — not run-to-run
variation, which is unmeasured here.
